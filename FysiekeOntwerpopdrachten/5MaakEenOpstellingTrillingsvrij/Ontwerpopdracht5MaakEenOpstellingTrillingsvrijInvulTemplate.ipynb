## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
f = 5 #Hz 
verhoudingA = 1/3
f_eigen = np.sqrt((f)**2)/(abs(1-(1/verhoudingA)))
print("kantel frequentie = ", f_eigen)
```

```python
# Voeg hier je plot in.
f = np.linspace(0,10,100)
verhouding = np.abs(1/(1-(f/f_eigen)**2))
plt.plot(f,verhouding)
plt.xscale("log")
plt.yscale("log")
plt.title("Bode plot")
plt.xlabel("Frequentie (Hz)")
plt.ylabel("Amplitude verhouding")
plt.show()

```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81 # m/s^2
delta_x = g/(2*np.pi*f_eigen)**2
print("De minimale delta x (in meters) is", np.round(delta_x,3))
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
m = np.linspace(0.1,2,20)
C = (m*g)/(delta_x)
plt.plot(m,C,"k.", label = "originele delta x")
plt.plot(m,2*C,"g.", label = "factor 2 delta x")
plt.plot(m,5*C,"b.", label = "factor 5 delta x")
plt.title("Plot massa tegen veerconstante")
plt.xlabel("Massa (kg)")
plt.ylabel("Veerconstante (N/m)")
plt.legend()
plt.show()
plt.savefig("Massa tegen Veerconstante")
```
