##### Vraag 1a Het potentiaal rond een puntlading

1. Bepaal welke index correspondeert met $z = 0$ m.
2. Plot de log van het potentiaal tegen de log van de straal.
3. Fit de log van het potentiaal met de log van de straal.
4. Print het exponent op een nette manier.

````python
# 1: Bepaal welk index correspondeert met z = 0 m

index_z0 = np.argmin(np.abs(z)) # takes minimum value with a distance to 0
print("De index die correspondeert met z = 0 m is", index_z0) # -1.77635684e-14 is ongeveer 0 (index = 50)

# 2: Plot de log van het potentiaal tegen de log van de straal

V_r = V[index_z0, :] # all r at this z 
r_nonzero = r[1:] # log is undefined at origin, so skip r = 0 (better for fit function)
V_nonzero = V_r[1:] # skip corresponding V

# loglog plot 

plt.figure(figsize=(6,4))
plt.loglog(r_nonzero, V_nonzero, marker='o', linestyle="none", color = "k")
plt.xlabel("$r$ (m)")
plt.ylabel("$V$ (V)")
plt.title("Log-Log plot van potentiaal tegenover straal")
plt.grid(True, which="both", ls="--")
plt.show()

# 3: Fit en plot de log van het potentiaal met de log van de straal.
from scipy.optimize import curve_fit

def fitfunctie(r, k, n): # define fit function (exponential)
    return k*r**n

popt, pcov = curve_fit(fitfunctie, r_nonzero, V_nonzero)
k_fit = popt[0] # k = 8.987742437988218e-05 (roughly 1/4*pi*epsilon)
n_fit = popt[1] # n = -1.0 

# fitted curve
V_fit = fitfunctie(r_nonzero, k_fit, n_fit)

plt.figure(figsize=(6,4))
plt.loglog(r_nonzero, V_nonzero, marker='o', linestyle="none", color = "k", label="Data")
plt.loglog(r_nonzero, V_fit, "-", color='b', label="Model")
plt.xlabel("$r$ (m)")
plt.ylabel("$V$ (V)")
plt.title("Log-Log plot van potentiaal tegenover straal")
plt.grid(True, which="both", ls="--")
plt.legend(loc="upper left")
plt.show()


# uncertainties 
uk_fit = np.sqrt(pcov[(0,0)]) # uncertainty of k_fit = -1.0e+00
un_fit = np.sqrt(pcov[(1,1)]) # uncertainty of n_fit = 2.0e-17

# 4: final answer exponent 
print(f"De exponent (n) is: {n_fit:.1e} ± {un_fit: .1e}") 


````

##### Vraag 1b Het elektrische veld rond een puntlading
1. Voer van vraag 1a 2 tot 4 uit voor het elektrische veld.

````python
# 1: Plot de log van het elektrisch veld tegen de log van de straal.

# extract E along r at z ≈ 0
normE_r = normE[index_z0, :]
r_skip = r_for_E[3:]             # skip r values for log-log and fit function
normE_skip = normE_r[3:]


# log-log plot of both components
plt.figure(figsize=(6,4))
plt.loglog(r_skip, normE_skip, 'o-', color='k', linestyle="none")
plt.xlabel("$r$ (m)")
plt.ylabel("$|E|$ (V/m)")
plt.title("Elektrisch veld tegenover straal")
plt.grid(True, which="both", ls="--", alpha=0.7)
plt.tight_layout()
plt.show()


# 2: Fit functie 
popt, pcov = curve_fit(fitfunctie, r_skip, np.abs(normE_skip), p0=[1e-4,-2])  # fit magnitude
k_fit = popt[0] 
n_fit = popt[1] 

# uncertainties
uk_fit = np.sqrt(pcov[0,0])
un_fit = np.sqrt(pcov[1,1])

E_fit = fitfunctie(r_skip, k_fit, n_fit)

plt.figure(figsize=(6,4))
plt.loglog(r_skip, np.abs(normE_skip), "o", color="k", label="data")
plt.loglog(r_skip, E_fit, "-", color="b", label= "model")
plt.xlabel("$r$ (m)")
plt.ylabel("$|E|$ (V/m)")
plt.title("Elektrisch veld tegenover straal incl. fit functie")
plt.grid(True, which="both", ls="--", alpha=0.7)
plt.legend(loc="upper right")
plt.tight_layout()
plt.show()


# 3: Final value exponent 
print(f"De exponent (n) is: {n_fit:.1e} ± {un_fit: .1e}")
````

De oppervlakte lading $\sigma(r)$ van een geleidende schijf met straal $R$ is niet uniform verdeeld over de schijf, maar wordt gegeven door (Problems 2.58 and 2.63 Griffiths 5th edition):
$$\sigma(r) = \frac{Q}{2 \pi R \sqrt{R^2 - r^2}}$$

1. Bereken de totale lading op de schijf met behulp van vraag 3a en 3b.

4. Plot in hetzelfde figuur de oppervlaktelading van een uniform geladen schijf met dezelfde lading $Q$.


````python
# bepaling van de totale lading op de schijf
Qtot = 2 * np.pi * disk_radius**2 * sigma0
print(f"De totale lading op de schijf is ongeveer {Qtot:.2e} C")

# bereken de oppervlakte lading in het midden van de schijf bij een geleidende verdeling en bij een uniforme verdeling
sigma0_geleidende_0 = Qtot / (2 * np.pi * disk_radius**2)
sigma0_uniform = Qtot / (np.pi * disk_radius**2)
print(f"De oppervlakte lading in het midden van de schijf bij een geleidende verdeling is ongeveer {sigma0_geleidende_0:.2e} C/m^2")
print(f"De oppervlakte lading in het midden van de schijf bij een uniforme verdeling is ongeveer {sigma0_uniform:.2e} C/m^2")
display(Math(r"\sigma_{\text{geleidende}} = \frac{1}{2}\sigma_{\text{uniform}}"))

# Plot de oppervlakte lading als functie van de straal voor een geleidende verdeling en vergelijk met de uniforme verdeling
r_range = np.linspace(0, disk_radius*0.99, 5000)
def sigma_geleidende(r):
    return Qtot / (2 * np.pi * disk_radius * np.sqrt(disk_radius**2 - r**2))

plt.figure()
plt.plot(r_range, sigma_geleidende(r_range))
plt.axhline(sigma0_uniform, color='r', linestyle='--', label="Uniforme lading")
plt.title("Oppervlakte lading als functie van de straal")
plt.xlabel("Straal (m)")
plt.ylabel("Oppervlakte lading (C/m^2)")
plt.ylim(0, 4*sigma0_uniform)
plt.show()
````

````python
# functies voor capaciteit van vlakke plaat en cilinder condensator
def vlakkeplaat(length, width, spacing_between_disks, epsilon_r):
    C = epsilon0 * epsilon_r * length * width / spacing_between_disks
    return C

def cilinder(lengte, binnenstraal, buitenstraal, epsilon_r):
    C = epsilon0 * epsilon_r * 2 * np.pi * lengte / np.log(buitenstraal / binnenstraal)
    return  C
````

$ C = \frac{\varepsilon A}{d}$

luchtlaag $\Rightarrow$ $C_{\mathrm{lucht}}  = \frac{\varepsilon_0 A}{d}$

kartonlaag $\Rightarrow$ $ C_{\mathrm{karton}} = \frac{\varepsilon_0 \varepsilon_r A}{d} $

$ \frac{1}{C} = \frac{1}{C_{\mathrm{lucht}}} + \frac{1}{C_{\mathrm{karton}}} $

$ \frac{1}{C} = \frac{d_{\mathrm{lucht}}}{\varepsilon_0 A} + \frac{d_{\mathrm{karton}}}{\varepsilon_0 \varepsilon_r A}$

$ C = \frac{\varepsilon_0 A}{d_{\mathrm{lucht}} + \frac{d_{\mathrm{karton}}}{\varepsilon_r}} $
